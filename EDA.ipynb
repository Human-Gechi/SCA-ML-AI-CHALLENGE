# **Exploratory Data Analysis (EDA)**

### **Study Population**  
This dataset includes information from Pakistan on standardized test scores, maternal and family demographics, household amenities, family structure, and socioeconomic indicators. It also contains maternal health and well-being measures relevant to postpartum outcomes.

### **Aim**  

**In an Abstract from the website of the National Institute of health (of the united states),Postpartum depression (PPD) affects around 10% of women👩‍🦰, or 1 in 7 women👩‍🦰, after giving birth. Undiagnosed PPD was observed among 50% of mothers. PPD has an unfavorable relationship with women’s functioning, marital and personal relationships, the quality of the mother-infant connection, and the social, behavioral, and cognitive development of children.**

The purpose of this exploratory analysis is to identify key patterns and relationships within the data that may influence postpartum depression as well as person or things postpartum depression affects.Insights gained will support the development of a predictive model for postpartum depression and inform potential intervention strategies.

```python
#Function for grouping hand depression scores at the baseline or start of the study
def baseline_depression(x):
    if x in range(0,8):
        return "No Depression"
    elif x in range(8, 14):
        return "Mild Depression"
    elif x in range(14, 19):
        return "Moderate Depression"
    elif x in range(19, 23):
        return "Severe Depression"
    else:
        return "Very Severe Depression"
```

```python
#function for depressed persond
def depressed(series):
    return (series == 1).sum()  #"yes"
```

```python
#function for non depressed persons
def not_depressed(series):
    return (series == 0).sum()
```

#### **Are moms in the poorest group more likely to be depressed than moms in the richest group, at any point in time (6 months, 1 year, or baseline)?**

The bins were explicitly defined because a duplicate error (100) occurred. Since the error message specified the bins, it became clear that the issue was with their definition. Therefore, the bins were explicitly stated in the subsequent cell to resolve the error.


```python
# Defining income quantile bins as well as labels
df = df.assign(
    income_quantile=pd.cut(
        df['income_hh'],
        bins=[0.0, 80.0, 100.0, 150.0, 800.0],
        labels=['Q1_Poorest', 'Q2', 'Q3', 'Q4_Richest'],
        include_lowest=True
    )
)

# Depression variables
depression_vars = ['depressed_6m', 'depressed_1y', 'dep_sample',"depressed"]

print("Depression rates per income quantile")
print("Percentages of mothers who are depressed")

# Storing comparison results
all_higher = []
rate_summary = {}

# Looping through each depression variable
for dep_var in depression_vars:
    rates = df.groupby('income_quantile', observed=False)[dep_var].mean() * 100
    q1_rate = rates['Q1_Poorest']
    q4_rate = rates['Q4_Richest']
    diff = q1_rate - q4_rate
    higher = q1_rate > q4_rate

    # Print results
    print(f"\n{dep_var}:")
    print(f"Q1_Poorest: {q1_rate:.1f}%")
    print(f"Q4_Richest: {q4_rate:.1f}%")
    print(f"Q1 higher than Q4? {'Yes' if higher else 'No'} (difference: {diff:+.1f}%)")

    # Store results
    all_higher.append(higher)
    rate_summary[dep_var] = (q1_rate, q4_rate, higher)

# Summary conclusion
overall_answer = "Yes" if any(all_higher) else "No"
print("\nSummary")
print(f"Do poorest households show higher depression rates overall? {overall_answer}")

# Evidence of the quantiles
print("\nEvidence")
for dep_var in depression_vars:
    q1, q4, higher = rate_summary[dep_var]
    print(f"{dep_var}: {q1:.1f}% vs {q4:.1f}% - {'Yes' if higher else 'No'}")
```

##### Do children of depressed mothers attain the expected mean child height and weight?

```python
#Converting child ages to z scores to compare the ages of kids and their expected ages
zscore_cols = {}

#For loop to loop through the list and calculat ethe z scores for their respective heights and weights
for age, wt_col, ht_col in [("6m", "c_wt_6m", "c_ht_6m"), ("1y", "c_wt_1y", "c_ht_1y")]:
    # Weight-for-age z-score (WAZ)
    mean_wt, std_wt = df[wt_col].mean(), df[wt_col].std()
    zscore_cols[f"ch_{age}_wt_z"] = (df[wt_col] - mean_wt) / std_wt

    # Height-for-age z-score (HAZ)
    mean_ht, std_ht = df[ht_col].mean(), df[ht_col].std()
    zscore_cols[f"ch_{age}_ht_z"] = (df[ht_col] - mean_ht) / std_ht

# Add all z-score columns to the dataframe at once
df = pd.concat([df, pd.DataFrame(zscore_cols, index=df.index)], axis=1)
```

# Child Growth and Maternal Depression

Comparing the mean growth of children with **depressed** and **non-depressed mothers** at 6 months and 1 year, using international growth standards:

- **HAZ (Height-for-Age Z-score):** how tall a child is compared to expected height for age.  
- **WAZ (Weight-for-Age Z-score):** how heavy a child is compared to expected weight for age.  
- **ch_wt_z and ch_ht_z:** the child's actual weight and height relative to their expected values.  

In z-scores:  
- **0** = exactly at the expected average.  
- **Negative** = below the expected average.  
- **Positive** = above the expected average.

## 1. Growth at 6 Months
- **All children** were already **slightly below the international standard** (both HAZ and WAZ were negative).  
- Children of **non-depressed mothers** had values **closer to 0**, meaning their weight and height were nearer the expected range.  
- Children of **depressed mothers** had **more negative scores**, showing they were lighter and shorter compared to both the standard and their peers.  

**Note:** By six months, a gap in growth had already formed — children of depressed mothers were falling behind.  

## 2. Growth at 1 Year
- **All children** fell further **below the international standard** (HAZ and WAZ became even more negative).  
- Children of **non-depressed mothers** still performed better, staying closer to expected growth levels.  
- While the children of depressed mothers had better mean z scores as opppsed to the standard mean z scores for their ages, Children of **depressed mothers** showed the **lowest scores** compared to their non depressed mother counterpart indicating greater deficits in both weight and height.  

**Note** By one year, the gap between children of depressed and non-depressed mothers widened, with depressed mothers’ children showing more glaring height and weight deficiences.

#### Overall Interpretation
- It is worth highlighting that the children of **non-depressed mothers** are doing *relatively better*, while children of **depressed mothers** are **further behind**.  
- This suggests that **maternal depression is linked to slower child growth** in both weight and height.
